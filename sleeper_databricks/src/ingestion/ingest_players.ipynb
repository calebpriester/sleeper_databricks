```python
from pyspark import pipelines as dp
import requests
import pandas as pd

@dp.table(
    name="players",
    comment="Raw NFL player data from Sleeper API"
)
def players():
    url = "https://api.sleeper.app/v1/players/nfl"
    response = requests.get(url)
    data = response.json()
    players = list(data.values())
    pdf = pd.DataFrame(players)
    # Convert columns with non-scalar types to string
    for col in pdf.columns:
        pdf[col] = pdf[col].apply(
            lambda x: str(x) if isinstance(x, (list, dict)) or pd.isnull(x) else x
        )
    return spark.createDataFrame(pdf)
```
